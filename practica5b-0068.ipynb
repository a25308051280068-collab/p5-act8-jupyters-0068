```python
# gastos diarios en un diccionario key, dia gastos
# values lista de datos y lista de gastos
datos = {
    "Dia": ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"],
    "Gasto": [50, 35, 80, 45, 60]
}
# dataframe tabla o marco de datos
df = pd.DataFrame(datos)
print("=== GASTOS SEMANALES ===")
print(df)
```

```python
print("=== El total de datos y el promedio de gastos ===")
print("Gasto total:", df["Gasto"].sum())
print("Gasto promedio:", df["Gasto"].mean())
print("\nDías con gasto mayor a $50:")
print(" key gasto value  [50, 35, 80, 45, 60]")
print(df[df["Gasto"] > 50])

print("+-+-+--+-+-+-+ Christian Garcia 0068-+-+-+-+-")
```
